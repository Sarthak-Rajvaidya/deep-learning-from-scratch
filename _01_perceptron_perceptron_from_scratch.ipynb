```python
x1 = 2
x2 = 3

print("x1 :",x1)
print("x2 :",x2)

w1 = 0.5
w2 = 0.8

print("w1 : ",w1)
print("w2 :",w2)

weighted_sum = x1*w1 + x2*w2

print("weighted sum : ",weighted_sum)

b = -2;

z = x1*w1 + x2*w2 + b

print("z : ",z)


def step_activation(z):
  if z >= 0:
    return 1
  else:
    return 0

  output = step_activation(z)

  print("Output : ",output)
```
