```python
x1 = 2
x2 = 3

print("x1 :",x1)
print("x2 :",x2)

w1 = 0.5
w2 = 0.8

print("w1 : ",w1)
print("w2 :",w2)

weighted_sum = x1*w1 + x2*w2

print("weighted sum : ",weighted_sum)

b = -2;

z = x1*w1 + x2*w2 + b

print("z : ",z)


def step_activation(z):
  if z >= 0:
    return 1
  else:
    return 0

  output = step_activation(z)

  print("Output : ",output)
```

```python
import math

def sigmoid(x):
  return 1 / (1+math.exp(-x))
```

```python
print(sigmoid(0))
print(sigmoid(2))
print(sigmoid(-2))
```

```python
def tanh(x):
  return (math.exp(-x)-math.exp(x)) / (math.exp(-x)+math.exp(x))
```

```python
print(tanh(0))
print(tanh(2))
print(tanh(-2))
```

```python
def relu(x):
  if x> 0:
    return x
  else:
    return 0
```

```python
print(relu(0))
print(relu(2))
print(relu(-2))
```
