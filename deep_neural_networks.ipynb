```python
import math
import random
```

```python
#2. Basic Perceptron

# A perceptron calculates: z= w1x1+b

def step_function(z):
    if z >= 0:
        return 1
    else:
        return 0

# Perceptron prediction

def perceptron_predict(x1, x2, w1, w2, b):
    z = w1 * x1 + w2 * x2 + b
    prediction = step_function(z)

    return prediction

x1 = 1
x2 = 1

w1 = 0.5
w2 = 0.5
b = -0.5

prediction = perceptron_predict(x1, x2, w1, w2, b)

print("Prediction:", prediction)
```

```python
# 3. Activation Functions

# Now move beyond the simple perceptron step function.

# Sigmoid

# σ(x)=1+e−x1​

def sigmoid(x):
    return 1 / (1 + math.exp(-x))

print("Sigmoid:")
print(sigmoid(-2))
print(sigmoid(0))
print(sigmoid(2))

# Tan h

def tanh(x):
    return (math.exp(x) - math.exp(-x)) / (
        math.exp(x) + math.exp(-x)
    )

print("Tanh:")
print(tanh(-2))
print(tanh(0))
print(tanh(2))

# ReLU

# ReLU(x)=max(0,x)

def relu(x):
    if x > 0:
        return x
    else:
        return 0
def relu(x):
    if x > 0:
        return x
    else:
        return 0
```

```python
# # 4. Activation Derivatives

# # These are important for backpropagation.

# # Sigmoid derivative

# σ
# # (x)=σ(x)(1−σ(x))

def sigmoid_derivative(x):
    s = sigmoid(x)
    return s * (1 - s)

# Tanh derivative

# tanh
# ′
# (x)=1−tanh
# 2
# (x)

def tanh_derivative(x):
    t = tanh(x)
    return 1 - t**2

#ReLU derivative

def relu_derivative(x):
    if x > 0:
        return 1
    else:
        return 0
print("Sigmoid derivative:", sigmoid_derivative(0))
print("Tanh derivative:", tanh_derivative(0))
print("ReLU derivative:", relu_derivative(-2))
print("ReLU derivative:", relu_derivative(2))
```

```python
# 5. Forward Propagation — Single Neuron

x = 2.0

w = 0.5
b = 0.0

z = w * x + b

a = sigmoid(z)

print("Weighted sum:", z)
print("Prediction:", a)
```

```python
# 6. Loss Function

# We'll use Mean Squared Error for one example

# L= (1/2(a-y)**2)

y = 1.0

loss = 0.5 * (a - y) ** 2

print("Loss:", loss)
```

```python
# 7. Backpropagation

dL_da = a - y

# Step 2

# For sigmoid:

da_dz = a * (1 - a)

dz_dw = x

dL_dw = dL_da * da_dz * dz_dw

print("dL/dw:", dL_dw)
```

```python
# 8. Bias Gradient

dz_db = 1

dL_db = dL_da * da_dz * dz_db

print("dL/db:", dL_db)
```

```python
# 9. Gradient Descent

# Now we actually update the parameters.

learning_rate = 0.1

w = w - learning_rate * dL_dw
b = b - learning_rate * dL_db

print("Updated weight:", w)
print("Updated bias:", b)
```

```python
import math

# -------------------------
# Training data
# -------------------------

x = 2.0
y = 1.0


# -------------------------
# Initial parameters
# -------------------------

w = 0.5
b = 0.0

learning_rate = 0.1


# -------------------------
# Activation function
# -------------------------

def sigmoid(x):
    return 1 / (1 + math.exp(-x))


# -------------------------
# Training
# -------------------------

for epoch in range(20):

    # Forward propagation

    z = w * x + b

    a = sigmoid(z)


    # Loss

    loss = 0.5 * (a - y) ** 2


    # Backpropagation

    dL_da = a - y

    da_dz = a * (1 - a)

    dL_dw = dL_da * da_dz * x

    dL_db = dL_da * da_dz


    # Gradient descent

    w = w - learning_rate * dL_dw

    b = b - learning_rate * dL_db


    print(
        f"Epoch {epoch + 1:2d} | "
        f"Prediction: {a:.4f} | "
        f"Loss: {loss:.4f} | "
        f"Weight: {w:.4f} | "
        f"Bias: {b:.4f}"
    )
```

```python
# 12. SGD From Scratch

X = [1, 2, 3, 4]
Y = [2, 4, 6, 8]

w = 0.0
b = 0.0

learning_rate = 0.01

epochs = 10


for epoch in range(epochs):

    for x, y in zip(X, Y):

        # -------------------------
        # Forward
        # -------------------------

        prediction = w * x + b

        # -------------------------
        # Loss
        # -------------------------

        loss = 0.5 * (prediction - y) ** 2

        # -------------------------
        # Backpropagation
        # -------------------------

        dL_dprediction = prediction - y

        dprediction_dw = x
        dprediction_db = 1

        dL_dw = dL_dprediction * dprediction_dw
        dL_db = dL_dprediction * dprediction_db

        # -------------------------
        # SGD update
        # -------------------------

        w = w - learning_rate * dL_dw
        b = b - learning_rate * dL_db


    print(
        f"Epoch {epoch + 1}: "
        f"w={w:.4f}, "
        f"b={b:.4f}"
    )
```

```python
# 13. Mini-Batch SGD From Scratch

X = [1, 2, 3, 4, 5, 6, 7, 8]
Y = [2, 4, 6, 8, 10, 12, 14, 16]

w = 0.0
b = 0.0

learning_rate = 0.01

batch_size = 2
epochs = 10
```

```python
for epoch in range(epochs):

    for start in range(0, len(X), batch_size):

        # -------------------------
        # Select mini-batch
        # -------------------------

        batch_X = X[start:start + batch_size]
        batch_Y = Y[start:start + batch_size]


        # Accumulate gradients

        total_dL_dw = 0
        total_dL_db = 0


        # -------------------------
        # Process batch
        # -------------------------

        for x, y in zip(batch_X, batch_Y):

            # Forward
            prediction = w * x + b

            # Gradient
            dL_dprediction = prediction - y

            dL_dw = dL_dprediction * x
            dL_db = dL_dprediction


            # Accumulate
            total_dL_dw += dL_dw
            total_dL_db += dL_db


        # -------------------------
        # Average gradients
        # -------------------------

        batch_length = len(batch_X)

        average_dL_dw = total_dL_dw / batch_length
        average_dL_db = total_dL_db / batch_length


        # -------------------------
        # Update parameters
        # -------------------------

        w = w - learning_rate * average_dL_dw

        b = b - learning_rate * average_dL_db


    print(
        f"Epoch {epoch + 1}: "
        f"w={w:.4f}, "
        f"b={b:.4f}"
    )
```
