```python
import math
import random
```

```python
#2. Basic Perceptron

# A perceptron calculates: z= w1x1+b

def step_function(z):
    if z >= 0:
        return 1
    else:
        return 0

# Perceptron prediction

def perceptron_predict(x1, x2, w1, w2, b):
    z = w1 * x1 + w2 * x2 + b
    prediction = step_function(z)

    return prediction

x1 = 1
x2 = 1

w1 = 0.5
w2 = 0.5
b = -0.5

prediction = perceptron_predict(x1, x2, w1, w2, b)

print("Prediction:", prediction)
```

```python
# 3. Activation Functions

# Now move beyond the simple perceptron step function.

# Sigmoid

# σ(x)=1+e−x1​

def sigmoid(x):
    return 1 / (1 + math.exp(-x))

print("Sigmoid:")
print(sigmoid(-2))
print(sigmoid(0))
print(sigmoid(2))

# Tan h

def tanh(x):
    return (math.exp(x) - math.exp(-x)) / (
        math.exp(x) + math.exp(-x)
    )

print("Tanh:")
print(tanh(-2))
print(tanh(0))
print(tanh(2))

# ReLU

# ReLU(x)=max(0,x)

def relu(x):
    if x > 0:
        return x
    else:
        return 0
def relu(x):
    if x > 0:
        return x
    else:
        return 0
```
